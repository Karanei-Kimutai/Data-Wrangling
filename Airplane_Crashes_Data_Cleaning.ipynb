5. Create a new dataframe from the raw/uncleaned data called
‘fatality_locations’ with the following columns: ‘Date’,
‘Location’, ‘Aboard’, ‘Fatalities’.


```python
fatality_locations=df[['Date', 'Location', 'Aboard', 'Fatalities']].copy()
```

6. Determine the date of the highest number of recorded fatalities from this new
dataframe.


```python
dateOfHighestFatalities= fatality_locations.loc[fatality_locations['Fatalities'].idxmax(), 'Date']
print(f"Date of Highest Fatalities: {dateOfHighestFatalities}")
```

7. Compare the number of passengers aboard to the number of fatalities per crash. Are
there any recorded crashes where there were no fatalities? If so, how many?


```python
# Compare the number of passengers aboard to the number of fatalities
fatality_locations['Survivors'] = fatality_locations['Aboard'] - fatality_locations['Fatalities']
# Display comparison for a couple of records
print("Comparison of Passengers Aboard vs Fatalities:")
print(fatality_locations[['Aboard', 'Fatalities', 'Survivors']].head(10))
```

```python
# Check recorded crashes where there were no fatalities
noFatalities=fatality_locations[fatality_locations['Fatalities'] == 0]
print(f"Number of crashes with zero fatalities: {len(noFatalities)}")
# Display those records
print(noFatalities)
```

8. In the ‘fatality_locations’ dataframe, the location column has data items
represented as ‘Region, U.S. State/Country’. Split this column to have two columns
of regions and states/countries.

```python
#fatality_locations[['Region', 'U.S. State/Country']]=fatality_locations['Location'].str.split(', ', n=1, expand=True)
# The n=1 parameter ensures it only splits on the first comma
# Display the original Location column alongside the newly created split columns
# Ran into an edge case where the U.S.State/Country column had values like Queens, New York meaning the split string had more than one comma and was possibly JFK, Queens, New York
# Will do a reverse split instead
fatality_locations[['Region', 'U.S. State/Country']] = fatality_locations['Location'].str.rsplit(', ', n=1, expand=True)
print(fatality_locations[['Location', 'Region', 'U.S. State/Country']].head(10))
```

9. Order the dataframe by the number of fatalities from highest to lowest and select the
first 100.


```python
top100Fatalities= fatality_locations.sort_values(by='Fatalities', ascending=False).head(100)
print("Top 100 Fatalities:")
print(top100Fatalities)
```

10.  Generate a pie chart that shows the distribution of the top 25 number of fatalities
per country/U.S. State. [Hint: charts can be generated from the seaborn library]

```python
top25Fatalities=top100Fatalities.head(25)
groupFatalitiesByState=top25Fatalities.groupby('U.S. State/Country')['Fatalities'].sum()

colors=sns.color_palette('pastel')[0:len(groupFatalitiesByState)]
plt.figure(figsize=(10,8))
plt.pie(groupFatalitiesByState, labels=groupFatalitiesByState.index, colors=colors, autopct='%1.1f%%')
plt.title('Distribution of Fatalities in Top 25 Crashes by Location')
plt.show()
```
